# Customer Data Cleaning & Analysis
A portfolio project demonstrating data cleaning, preprocessing, exploratory analysis, and visualization using Python.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('../data/raw_data.csv')
df.head()
```

## Data Inspection

```python
print('Shape:', df.shape)
print('\nMissing values:')
print(df.isnull().sum())
print('\nDuplicate rows:', df.duplicated().sum())
df.describe(include='all')
```

## Data Cleaning

```python
cleaned = df.drop_duplicates().copy()
cleaned['Age'] = cleaned['Age'].fillna(cleaned['Age'].median())
cleaned['Payment_Method'] = cleaned['Payment_Method'].fillna('Unknown')
cleaned['Date'] = pd.to_datetime(cleaned['Date'])
cleaned = cleaned[(cleaned['Age'] >= 18) & (cleaned['Age'] <= 80)]
price_limit = cleaned['Unit_Price'].quantile(0.99)
cleaned = cleaned[cleaned['Unit_Price'] <= price_limit]
cleaned['Total_Sales'] = (cleaned['Quantity'] * cleaned['Unit_Price']).round(2)
cleaned.head()
```

## Analysis

```python
sales_by_category = cleaned.groupby('Category')['Total_Sales'].sum().sort_values(ascending=False)
sales_by_region = cleaned.groupby('Region')['Total_Sales'].sum().sort_values(ascending=False)
display(sales_by_category)
display(sales_by_region)
```

```python
plt.figure(figsize=(9,5))
sales_by_category.plot(kind='bar')
plt.title('Total Sales by Category')
plt.xlabel('Category'); plt.ylabel('Total Sales')
plt.xticks(rotation=30); plt.tight_layout(); plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.hist(cleaned['Total_Sales'], bins=30)
plt.title('Sales Distribution')
plt.xlabel('Total Sales'); plt.ylabel('Frequency')
plt.tight_layout(); plt.show()
```

```python
cleaned.to_csv('../data/cleaned_data.csv', index=False)
print('Cleaned dataset saved successfully.')
```
